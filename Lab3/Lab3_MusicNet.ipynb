## Dataset Analysis: musicnet_metadata.csv

## 1. Import Libraries

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (6,4)
```

## 2. Load Dataset

```
df = pd.read_csv("/mnt/data/musicnet_metadata.csv")
df.head()
```

## 3. Dataset Information

```
df.info()
```

## 4. Descriptive Statistics

```
df.describe()
```

## 5. Missing Values

```
df.isnull().sum()
```

## 6. Correlation Matrix

```
numeric_df = df.select_dtypes(include=[np.number])
numeric_df.corr()
```

## 7. Histogram of First Numeric Column

```
if len(numeric_df.columns) > 0:
    col = numeric_df.columns[0]
    plt.figure()
    plt.hist(numeric_df[col].dropna())
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
else:
    print("No numeric columns available for histogram.")
```

## 8. Boxplot of First Numeric Column

```
if len(numeric_df.columns) > 0:
    col = numeric_df.columns[0]
    plt.figure()
    plt.boxplot(numeric_df[col].dropna())
    plt.title(f"Boxplot of {col}")
    plt.show()
else:
    print("No numeric columns available for boxplot.")
```

## 9. Conclusion
Correlation analysis provided insight into relationships between numeric features. Histogram and boxplot visualizations highlighted distribution shape, spread, and potential outliers..
