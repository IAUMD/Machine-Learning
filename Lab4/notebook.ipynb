```
# 1. Import libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

pd.set_option("display.max_columns", None)

# 2. Load dataset
df = pd.read_csv("vehicles.csv", low_memory=False)

df.head()
```

```
# 3. Data Quality Assessment

# 3.1 Check data types
df.dtypes
```

```
# 3.2 Convert incorrect data types
df["createdOn"] = pd.to_datetime(df["createdOn"], errors="coerce")
df["modifiedOn"] = pd.to_datetime(df["modifiedOn"], errors="coerce")

df.dtypes[["createdOn", "modifiedOn"]]
```

```
# 4. Handling Missing Values

# 4.1 Detect missing values
df.isna().sum()
```

```
# 4.2 Remove records with missing values
df_removed = df.dropna()

print("Original shape:", df.shape)
print("After removing missing:", df_removed.shape)
```

```
# 4.3 Mean imputation for numeric columns
df_mean = df.copy()
numeric_cols = df_mean.select_dtypes(include=["int64", "float64"]).columns
df_mean[numeric_cols] = df_mean[numeric_cols].fillna(df_mean[numeric_cols].mean())

df_mean.isna().sum().sum()
```

```
# 4.4 Median imputation for numeric columns
df_median = df.copy()
df_median[numeric_cols] = df_median[numeric_cols].fillna(df_median[numeric_cols].median())

df_median.isna().sum().sum()
```

```
# 5. Handling Outliers using IQR (example on city08)

Q1 = df["city08"].quantile(0.25)
Q3 = df["city08"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["city08"] < lower) | (df["city08"] > upper)]
print("Number of outliers:", outliers.shape[0])

df_no_outliers = df[(df["city08"] >= lower) & (df["city08"] <= upper)]
print("After removing outliers:", df_no_outliers.shape)
```

```
# 6. Capping Outliers (Percentile method)

lower_cap = df["city08"].quantile(0.05)
upper_cap = df["city08"].quantile(0.95)

df_capped = df.copy()
df_capped["city08"] = df_capped["city08"].clip(lower_cap, upper_cap)

df_capped["city08"].describe()
```

```
# 7. Min-Max Normalization (selected numeric columns)

scaler = MinMaxScaler()
cols_to_scale = ["city08", "highway08", "comb08"]

df_scaled = df[cols_to_scale].copy()
df_scaled[cols_to_scale] = scaler.fit_transform(df_scaled)

df_scaled.head()
```
